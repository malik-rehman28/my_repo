### 1) My lane (or freestyle) and why
**Lane Choice:** Lane 2: Refresh / Content Opportunity Scoring.
**Why:** Building a pipeline that transforms raw historical search signals into predictive insights offers immediate operational value. Instead of manually parsing dashboards to find decaying content, this lane creates a ranked queue. It forces a clear comparison between a transparent baseline rule (e.g., "stale visible page") and a learned ML model.

### 2) The question: decision, action, cost of a wrong call
* **The Search Question:** Which specific pages should our team review first for a content refresh or expansion?
* **Unit of Analysis:** A single anonymized content item (`content_id`).
* **The Output:** A ranked review queue featuring a final opportunity score and specific reason codes.
* **The Action:** A content editor reviews the top-K pages to rewrite titles, improve intent match, update stale information, or merge decaying pages.
* **Cost of a Wrong Call:** 
    * *False Positive:* Wasting limited editorial capacity reviewing a page experiencing natural seasonality.
    * *False Negative:* Missing a high-value "champion" page that is actively decaying.
* **Why ML Helps:** Simple thresholds treat all drops equally. A supervised model weighs multiple interactions simultaneously to separate true decline risks from normal fluctuations.

```python
import pandas as pd

# Load the data (the file is now in the current directory)
df = pd.read_csv('content_refresh_anonymized.csv')

total_pages = len(df)
decline_count = len(df[df['trend_direction'] == 'down'])
decline_rate = (decline_count / total_pages) * 100
high_volume = len(df[df['impressions_90d'] >= 500])

print(f"1. Total Pages: {total_pages:,}")
print(f"2. Decline Rate: {decline_rate:.1f}%")
print(f"3. Actionable Pages (>500 impressions): {high_volume:,}")
```

### 4) Careful words: what I can and can't claim
* **What I can claim:** I can claim that specific observable signals correlate with content decline, and that a model can prioritize these at-risk pages more effectively than a static rule.
* **What I cannot claim:** I cannot claim that updating the content will guarantee a traffic recovery, as this data is observational and lacks a causal experiment.
