1) My lane as an ML task

Task Type: Ranking / Scoring.

Task Definition: In this lane, the system needs to score and rank anonymized content items so that we can prioritize which specific pages our team should review first for a content refresh or expansion

2) Target or proxy


Target Variable: The trend_direction (specifically predicting the 'down' class) or an aggregated opportunity score.

Why this target: We are predicting this because identifying actively decaying pages allows us to find content that needs intervention, rather than pages experiencing normal seasonality.

3) Success metric


Metric: Precision@K (measuring the relevance of the top-ranked pages) or F1-Score.

Why this metric: This metric is appropriate because we must balance the cost of a false positive (wasting limited editorial capacity on normal fluctuations) against the cost of a false negative (missing a high-value "champion" page that is actively decaying).

4) The unit of analysis, as a real dataframe

Unit of Analysis: A single anonymized content item (content_id).

5) Why ML beats a fixed rule here

A fixed rule (like a transparent baseline rule of "stale visible page") fails here because simple thresholds treat all drops equally. Machine Learning is required because a supervised model can weigh multiple interactions simultaneously to separate true decline risks from normal, natural fluctuations.

6) Self-check

The Content Action: When the model outputs a ranked review queue featuring a final opportunity score, a content editor will review the top-K pages. They will then take action to rewrite titles, improve intent match, update stale information, or merge the decaying pages.
